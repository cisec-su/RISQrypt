```python
PROJ_NAME = "dilithium_sign_ttest"
LABEL = ""
PRNG_OFF = True
N = 1000 if PRNG_OFF else 100000
OFFSET_ARR = [(i, range(120000)) for i in range(0, 1320000, 120000)]
```

```python
import sys
sys.path.insert(0, "../")
from host import ttest_analysis
%matplotlib widget
```

```python
ttest_analysis_arr = [ttest_analysis.TTestAnalysis(proj_name=PROJ_NAME, label=LABEL, offset=offset[0], N=N, prng_off=PRNG_OFF, init_ths=False) for offset in OFFSET_ARR]
```

```python
for ttest in ttest_analysis_arr:
    ttest.load_ttest_results(load_means=True)
```

```python
import numpy as np

ttest_analysis_arr[0].mean_trace_1 = ttest_analysis_arr[0].mean_trace_1[OFFSET_ARR[0][1]]
ttest_analysis_arr[0].mean_trace_0 = ttest_analysis_arr[0].mean_trace_0[OFFSET_ARR[0][1]]

for i in range(1, len(ttest_analysis_arr)):
    ttest_analysis_arr[0].mean_trace_1 = np.concatenate((ttest_analysis_arr[0].mean_trace_1, ttest_analysis_arr[i].mean_trace_1[OFFSET_ARR[i][1]]), axis=0)
    ttest_analysis_arr[0].mean_trace_0 = np.concatenate((ttest_analysis_arr[0].mean_trace_0, ttest_analysis_arr[i].mean_trace_0[OFFSET_ARR[i][1]]), axis=0)
```

```python
from matplotlib import pyplot as plt
%matplotlib widget

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(ttest_analysis_arr[0].mean_trace_1, label="Mean Trace 1")
ax.plot(ttest_analysis_arr[0].mean_trace_0, label="Mean Trace 0")
```

```python
import numpy as np

ttest_analysis_arr[0].ttest.result = ttest_analysis_arr[0].ttest.result[OFFSET_ARR[0][1]]

for i in range(1, len(ttest_analysis_arr)):
    ttest_analysis_arr[0].ttest.result = np.concatenate((ttest_analysis_arr[0].ttest.result, ttest_analysis_arr[i].ttest.result[OFFSET_ARR[i][1]]), axis=0)
   
```

```python
from matplotlib import pyplot as plt
import numpy as np

%matplotlib widget
fig, ax = plt.subplots()
ttest_analysis_arr[0].plot_ttest(ax=ax, color='black', plt_show=False, put_title=False, put_label=True, label_fontsize=18, hline=4.5, ylimit=6 if not PRNG_OFF else None)
ax.set_xlim(0, 1215000)
ax.set_ylim(0, np.abs(ttest_analysis_arr[0].ttest.result).max() * 1.1 if PRNG_OFF else 6)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
fig.tight_layout()
```
